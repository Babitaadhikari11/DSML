## Ordinary Least Squares with SimpleLR

```python
import pandas as pd
import numpy as np
```

```python
advertisement_df = pd.read_csv("Advertising.csv", index_col = 0)
```

```python
advertisement_df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import matplotlib as mpl
from matplotlib import pyplot as plt

fig = plt.figure(figsize=(15,4))
gs = mpl.gridspec.GridSpec(1,3)

# Plot of sales vs TV
ax = fig.add_subplot(gs[0])
ax.scatter(advertisement_df["TV"], advertisement_df["Sales"], color="red", marker=".")
ax.set_xlabel("TV")
ax.set_ylabel("sales")

# Plot of sales vs radio
ax = fig.add_subplot(gs[1])
ax.scatter(advertisement_df["Radio"], advertisement_df["Sales"], color="green", marker=".")
ax.set_xlabel("radio")
ax.set_ylabel("sales")

# Plot of sales vs newspaper
ax = fig.add_subplot(gs[2])
ax.scatter(advertisement_df["Newspaper"], advertisement_df["Sales"], color="blue", marker=".")
ax.set_xlabel("newspaper")
ax.set_ylabel("sales")

plt.show()
```

```python
X = advertisement_df.TV.to_numpy()
```

```python
y = advertisement_df.Sales.to_numpy()
```

```python
# calculaton of beta_1
x_mean = X.mean()
y_mean =  y.mean()
```

```python
beta_1 = np.sum((X - x_mean)* (y - y_mean)) / np.sum(np.power((X - x_mean), 2))
```

```python
print(beta_1)
```

```python
# calculation of beta_0
beta_0 = y_mean - (beta_1 * x_mean)
```

```python
print(beta_0)
```

```python
# best fit line equation
print(f" y = {beta_0:.3f} + {beta_1:.3f} x")
```

```python
# calculating SSE
y_hat = beta_0 + (beta_1 * X)
```

```python
SSE = np.sum((y - y_hat) ** 2)
```

```python
print(SSE)
```

```python
plt.scatter(X, y,color ="green")
plt.plot(X, beta_0+ (beta_1 * X), color = "red")
```

## from SK Learn

```python
from sklearn.linear_model import LinearRegression
```

```python
X = advertisement_df.TV.to_numpy()
y = advertisement_df.Sales.to_numpy()
linear_reg = LinearRegression()
```

```python
linear_reg.fit(X.reshape(-1,1), y.reshape(-1,1))
```

```python
# beta_0
linear_reg.intercept_
```

```python
# beta_1
linear_reg.coef_
```

```python
y_cap = linear_reg.predict(X.reshape(-1,1))
```

```python
y_cap
```

```python
SSE = np.sum(np.power((y - y_cap.flatten()),2))
```

```python
print(SSE)
```
