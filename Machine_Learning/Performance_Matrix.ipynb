```python
import pandas as pd
import numpy as np
```

```python
advertisement_df = pd.read_csv("Advertising.csv", index_col = 0)
```

```python
advertisement_df.head()
```

```python
X = advertisement_df[["TV"]].to_numpy()
y = advertisement_df[["Sales"]].to_numpy()
```

## train and test

```python
from sklearn.model_selection import train_test_split
```

```python
# while passing 2 arrays, we get 4 returned list for both X train, X_Split and Y_train and Y_Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state=999)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lin_regression = LinearRegression()
```

```python
lin_regression.fit(X_train, y_train)
```

```python
# beta_0
lin_regression.intercept_
```

```python
# beta_1
lin_regression.coef_
```

```python
y_cap = lin_regression.predict(X_train)
```

```python
y_cap
```

```python
SSE = np.sum(np.power((y_train.flatten() - y_cap.flatten()),2))
```

```python
SSE
```

## Evaulation

```python
from sklearn.metrics import max_error, mean_absolute_error, mean_squared_error, r2_score
```

```python
y_prediction = lin_regression.predict(X_test)
```

```python
max_error(y_test, y_prediction)
```

```python
mean_absolute_error(y_test, y_prediction)
```

```python
mean_squared_error(y_test, y_prediction)
```

# R2 coefficient
#### indicates percentage of variance, measure reltioship between your model and dependent variable on a scale of  0 - 100%

```python
r2_score(y_test, y_prediction) 
```
