```python
class Person:
    """
    This is documentation
    """

    # init as magical method
    def __init__(self, name, age):
        self.name =  name
        self.age = age
    
    def __str__(self):
        return f"{self.name} / {self.age}"

    def __repr__(self):
        return f"Person({self.name} , {self.age})"
    
    def __len__(self):
        return self.age
    
    def __call__(self, gender = "female"): #this method can take parameters
        return f"{self.name} is a {gender}"
```

```python
person = Person("Babita", 22)
print(person) 
```

```python

class Point:

    def __init__(self, x: int | float, y: int | float):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def __len__(self):
        return 2
    
    def quardent(self):
        if not self.x and self.y == 0:
            return "origin"

        quardent_mapping = {
            (True, True) : "first",
            (False, True) : "Second",
            (False, False) : "Third",
            (True, False) : "Fourth"
        }

        return quardent_mapping.get((self.x > 0, self.y>0))


    # operator overloading
    def __add__(self, other): #self = point , other for another object addition
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)
    
    def __mul__(self, others):
        new_x = self.x * others.x
        new_y = self.y * others.y
        return Point(new_x, new_y)
```

```python
point1 = Point(4,5)
point2 = Point(2,1)
```

```python
p4 = point1 * point2
```

```python
print(p4)
```

```python
point3 = point1 + point2
```
