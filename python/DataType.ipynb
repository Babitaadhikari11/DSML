```python
# basic set operations
A={1,2,3,4,5}
B={3,5,7,9,11}
```

```python
aUb=A.union(B) #using operator A |B
aUb
```

```python
aNb=A.intersection(B)
aNb
```

```python
a_b=A.difference(B)
a_b
```

```python
# symmetric difference, remove intersecting element
symmetric_difference=A.symmetric_difference(B) #symbolic representation: a^b
symmetric_difference
```

```python
# subset,
A.issubset(B) #operator representation A<=B
```

```python
# proper subset
B < A
```

```python
# disjoint
B.isdisjoint(A)
```
