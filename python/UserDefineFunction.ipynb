```python
def calculate_area(shape:str,**kwars):
    area=0
    if shape=="square":
        area=kwars.get("length",0)**2
    elif shape=="rectangle":
        if "length" in kwars.keys() and "breadth" in kwars.keys():
             area=kwars.get("length",0)*kwars.get("breadth",0)
    else:
        print(f"{shape} is not supported")
    return area
```

```python
calculate_area("rectangle",length=3,breadth=2)
```

```python
# calculate standard deviation
def standard_dev(*args):
    '''calculate std from given numbers
    parameter
        *numbers:int or float
        any amount of int or float values
    Returns

        Float: SD
    '''
    mean=sum(args)/len(args)
    numerator=sum((arg-mean)**2 for arg in args)
    std_deviation= (numerator/len(args))**0.5
    return round(std_deviation,2)
```

```python
standard_dev(1,2,3,4,55)
```

recursive function

```python
# two cases- base case and recursive case
def fib(n):
    # base case-stopping point
    if n==0:
        return 0
    elif n == 1:
        return 1
    else:
        # recursive case
        return fib(n-1)+fib(n-2)

fib(8)
```

```python
# memoizing recursive algorithm
cache={0:0,1:1}
def fib(n):
    # base case-stopping point
    if n in cache:
        return cache[n]
    else:
        # recursive case
        res= fib(n-1)+fib(n-2)
        cache[n]=res
        return res

fib(8)
```

```python
# factorical
def factorial(numbers):
    # base case
    if numbers == 1:
        return 1
    else:
        result = factorial(numbers-1)*numbers
        return result
factorial(5)
```

```python
# memoization
# factorical
cache = [1,1]


def factorial(numbers):
    # base case
    if numbers < len(cache):
     return cache[numbers]
    
    else:
        result = factorial(numbers-1) * numbers
        return result
    
factorial(5)
```
