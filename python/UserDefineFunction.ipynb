```python
# *args, **kwargs
def print_all(*args): #take multiple arguments
    print("positional arguments")
    print(args)
```

```python
print_all("a",1,"b",True)
```

```python
def sum_all(*args):
    return sum(args)
```

```python
sum_all(1,2,3,4,5,6)
```

```python
# keywords arguments
def keyword_arguments(**kwargs):
    print("keyword arguments")
    print(kwargs)
```

```python
keyword_arguments(num1=2,num2=3)
```

```python
def calculate_area(shape:str,**kwars):
    area=0
    if shape=="square":
        area=kwars.get("length",0)**2
    elif shape=="rectangle":
        if "length" in kwars.keys() and "breadth" in kwars.keys():
             area=kwars.get("length",0)*kwars.get("breadth",0)
    else:
        print(f"{shape} is not supported")
    return area
```

```python
calculate_area("rectangle",length=3,breadth=2)
```

```python
# calculate standard deviation
def standard_dev(*args):
    '''calculate std from given numbers
    parameter
        *numbers:int or float
        any amount of int or float values
    Returns

        Float: SD
    '''
    mean=sum(args)/len(args)
    numerator=sum((arg-mean)**2 for arg in args)
    std_deviation= (numerator/len(args))**0.5
    return round(std_deviation,2)
```

```python
standard_dev(1,2,3,4,55)
```

```python
# lambda function
power_up=lambda base,power:base**power
power_up(4,2)
```

```python
def cuboid(l,b,h)->float:
    area=2*(l*b+b*h+l*h)
    return area
```

```python
cuboid(2,3,4)
```

```python
cuboid_area = lambda l, b, h:2*(l*b + b*h + l*h)
cuboid_area(20,22,5)
```
