generating large sequence of values

```python
# used yield
def number_generator(start, end):
    for num in range(start, end+1):
        yield num
```

```python
# step 2: create generator object
gen = number_generator(4,12)
```

```python
print(gen)
```

```python
# use generator object to generate values
next(gen)
```

```python
# using loop
for num in gen:
    print(num)
```

```python
def is_prime(n):

        if n < 2:
            return False
    
        for i in range(2, int(n**0.5) + 1):
        
            if n % i == 0:
                return False

        return True
```

```python
def prime_generator(limit = 10):
    count, start = 1, 1

    while count <= limit:
        start += 1
        if is_prime(start):
            count += 1
            yield start
        start += 1
```

```python
prime_gen =  prime_generator(6)
```

```python
list(prime_gen)
```
