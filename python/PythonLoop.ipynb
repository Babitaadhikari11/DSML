Break and Continue

```python
# Break
i=0
while i<=10:
    print(i)
    i+=1
    if i==4:
        break
```

```python
# continue, this just skip the certain point 

i=0
while i<=10:
    i+=1
    if i==4:
        continue
    print(i)
    
   
    
```

```python
# fibonnaci series
num1=0
num2=1
i=0
while i<=10:
    print(num1)
    c=num1+num2
    num1=num2
    num2= c
    i+=1
```

```python
# using for loop
a=0
b=1
for _ in range(10):
    print(a)
    c=a+b
    a=b
    b=c
```

comprehension, way to create lists, dictionaries


```python
squared=[]
for i in range(1,11):
    squared.append(i**2)
print(squared)
```

```python
squared=[i**2 for i in range(1,11) if i%2==0] #only works for single blocks
print(squared)
```

```python
# dictionary comprehension
quared={i:i**2 for i in range(1,11) if i%2==0} #only works for single blocks
print(squared)
```

```python
# nested comprehension
adjectives=["red","big","heavy"]
fruits=["apple","banana","mango"]
pair=[f"{a} - {f}" for a in adjectives for f in fruits]
pair
```

```python
# nested while loop
i=0
while i<10:
    i+=1
    j=0
    while j<10:
        j+=1
        print(i,j)
```

```python
sent = "Although she was extremely tired after working a long shift at the local hospital, Maria still managed to stop by the grocery store, pick up fresh vegetables for dinner, and help her younger brother with his math homework."
 
# O/P : {words : len(word) } , if length of word is > 4
output=sent.split()
output
result_dict={}
for word in output:
    if len(word)>4:
        result_dict[word]=len(word)
```
