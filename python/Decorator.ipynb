Implementation

```python
def decorator(func): #function as argument or parameter

    # inner function
    def wrapper():
        print("before function call")
        print(func().upper()) #actual function call
        print("after function call")
    
    return wrapper # function as argument, return as function 
```

```python
@decorator
def greet():
    return "hello"
```

```python
greet()
```

```python
# get time decorator implementation
def timeit(func):

    def wrapper(email):
        start_time =  time.time()
        res = func(email)
        print("Total time taken : ", int(time.time()- start_time), "seconds")
        return res
    return wrapper
```

Getter - get value of an attribute - we use property decorator

```python
class Person:
    
    def __init__(self, name):
        self.full_name =  name

    # making normal method as getter attribute
    @property
    def name(self):
        print("Getting name ")
        return self.full_name
```

```python
obj = Person("Ram Thapa")
```

```python
obj.name 
```

```python
import uuid


class Person:
    # making private
    def __init__(self, name):
        self.__full_name =  name

    # making normal method as getter attribute
    @property
    def name(self):
        print("Getting ID ")
        return f"{self.__full_name[-3]}_{str(uuid.uuid1())[:8]}"
```

```python
obj = Person("Hhi")
```

```python
obj.name
```

Setter - set the value of an attribute - can be used with @property decorator with @setter decorator

```python
import uuid


class Person:
    # making private
    def __init__(self, name):
        self.__full_name =  name

    # making normal method as getter attribute
    @property
    def name(self):
        print("Getting name ")
        return self.__full_name
    
    # setter method, can take input
    @name.setter
    def name(self, value):
        print("Setting name ")
        self.__full_name =  value
```

```python
obj1 = Person("Jackson")
```

```python
# set name to jackson
obj1.name
```

```python
# setter method change name to new name
obj1.name = "Ram"
```

```python
# getter attribute changes its value now to new value get from setter
obj1.name
```

Deleter - delete an attribue, using del

```python
import uuid


class Person:
    # making private
    def __init__(self, name):
        self.__full_name =  name

    # making normal method as getter attribute
    @property
    def name(self):
        print("Getting name ")
        return self.__full_name
    
    # setter method, can take input
    @name.setter
    def name(self, value):
        print("Setting name ")
        self.__full_name =  value

    @name.deleter
    def name(self):
        print("Deleting name")
        # deleting to empty value
        self.__full_name = ""
        # deleting the object
        # del self.__full_name
```

```python
person = Person("anmol kc")
```

```python
person.name
```

```python
person.name = "the rock"
```

```python
person.name
```

```python
# removing
del person.name
```

```python
# now empty 
person.name
```
