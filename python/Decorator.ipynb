Implementation

```python
def decorator(func): #function as argument or parameter

    # inner function
    def wrapper():
        print("before function call")
        print(func().upper()) #actual function call
        print("after function call")
    
    return wrapper # function as argument, return as function 
```

```python
@decorator
def greet():
    return "hello"
```

```python
greet()
```

```python
import time
import uuid
```

```python
# get time decorator implementation
def timeit(func):

    def wrapper(email):
        start_time =  time.time()
        res = func(email)
        print("Total time taken : ", int(time.time()- start_time), "seconds")
        return res
    return wrapper
```

```python
@timeit
def get_reset_code(email):
    return str(uuid.uuid1())[:8]
```

```python
get_reset_code("jhon@gmail.com")
```
