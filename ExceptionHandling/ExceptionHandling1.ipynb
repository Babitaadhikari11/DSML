```python
try:
    num = int(input("Enter a number : "))
    print(6/num)
except ValueError as e:
    # print(e)
    # print("Invalid input")
    print(0)
```

```python
# multiple except statement
try:
    num = int(input("Enter a number : "))
    print(6/num)

except ValueError as e:
    # print(e)
    print("Invalid input")
    print(0)

# handling zero division error
except ZeroDivisionError as e:
    print("cannot divide by zero")
    print(0)
    
# if any kind of error not defined appear we can use Exception base class directly
except Exception as e:
    print("unknown Error", e)
    print(0)

print("End of program")
```

```python
# single except block to handle multiple types of exceptions
value = "abv"
try:
    print(int(value))

except (ValueError, TypeError) as e:
    print(f"{value} - Cannot convert to int", e)
```

```python
# try except in user defined class
class InvalidAgeException(Exception):
    """ Raise whrn input age is less than 18 """
    pass
```

```python
def check_age(age):
    
        if age >= 18:
            print("Good to go")
        else:
            raise InvalidAgeException("Must be an adult to buy liquor, age > 18")
```

```python
try:
    check_age(
        int(input("Enter age: "))
    )

except InvalidAgeException as e:
    print("transaction cancelled: customer age under 18")

except(ValueError, TypeError) as e:
    print("Invalid age - transaction cancelled")
```

```python
# else block
try:
    x = float(input("Number: "))
    y = int(input("Number: "))
    result =  x/y
except ValueError as e:
    print("Invalid input - ", e)
    result =  1
    
# only runs when try is successful
else:
    print("Final Result", result)

```

```python
# finally block - runs anyway either try runs or except runs or fail

# else block
try:
    x = float(input("Number: "))
    y = int(input("Number: "))
    result =  x/y
except ValueError as e:
    print("Invalid input - ", e)
    result =  1
    
# only runs when try is successful
else:
    print("Final Result", result)

finally:
    print("End of file")
```
