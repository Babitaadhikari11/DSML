```python
# ATM 
class InsufficientBalanceException(Exception):
    """Custom exception for insufficient balance."""
    pass

balance = 2560
def get_cash(withdraw_amt):
    if balance >= withdraw_amt:
        print("Withdrawn Successful")
        return f"New Balance {balance - withdraw_amt}"
    else:
        raise InsufficientBalanceException("Insufficient balance")
```

```python
get_cash(3000)
```

Question 1 — Age Validator

Write a Python function called check_age(age) that:

Checks whether age is an integer using isinstance().
If it is an integer, print "Valid age: <age>".
If it is not an integer, raise a TypeError with the message "Age must be an integer".

```python
input_age =  int(input("Enter a age in integer:  "))
def check_age(input_age):
 
    if isinstance (input_age, int):
        print("age: ", input_age)
    else:
        raise ValueError("Age must be integer", type(input_age))
```

```python
check_age(input_age)
```

Question 2 — A little harder

Create a function called square(number).

Requirements:

Check whether number is an int using isinstance().
If it is an integer, print its square.
Otherwise, raise a TypeError that also shows the actual type received.

```python
def square(number):
    if isinstance(number, int):
        square_num = number**2
        print(f" Square of the number is {square_num} ")
    else:
        raise TypeError(f"The actual number is {number}" )
```

```python
square(2)
```

Write a Python program that:

Takes the user's age as input.
If the age is less than 18, raise AgeNotValidError.
Display the error message:
Age must be 18 or above
Otherwise, print:
You are eligible
Handle your custom exception using try and except.
At the end, print:
Program completed

```python
class AgeNotValidError(Exception):
    pass

age =  int(input("Enter age :"))
try:
    if age < 18:
        raise AgeNotValidError("Age must be above 18")
    else:
     print("You are eligible")
except AgeNotValidError as e:
   print(e)
print("program completed")
```
