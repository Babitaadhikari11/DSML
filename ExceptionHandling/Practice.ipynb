Using else block
Write a program that divides two numbers. Print "Division successful" only if no exception occurs.

Expected concepts: try–except–else

```python
def division(a, b):
    try:
        result =  a/b
    except ZeroDivisionError:
        print("denominator must be non zero")
        print(0)
    else:
        print("division successfull")
        print("the result is : ", result)
```

```python
division(2,2)
```

Create a function safe_divide(a, b) that:

Returns division result if valid
Returns None if an exception occurs
Prints the error message

```python
def safe_divide(a,b):
    try: 
        result = a/b
        return result
    except ZeroDivisionError:
        print("enter non zero number in denominator")
        return None
    
```

Raising exceptions in functions
Create a function withdraw(balance, amount):

Raise ValueError if amount is greater than balance
Return remaining balance otherwise

```python
def withdraw(balance, amount):
    
    try:

        if amount > balance:
            raise ValueError("insufficient balance")

        else:
            remained_balance = balance - amount
            print("the total amount is now :", remained_balance)

    except ValueError as e:
        print(e)
```

```python
withdraw(2000,3000)
```

Raise TypeError manually
Write a function add_numbers(a, b):

Raise TypeError if inputs are not integers
Return sum otherwise

```python
from typing import Type


def add_numbers(a,b):
    try:
        if not isinstance(a,int) and not isinstance(b, int):
            raise TypeError("must be an integer")
        else:
            result =  a + b
            return result
    
    except TypeError as e:
        print(e)
        print(0)
print("end")
```

```python
add_numbers(2,2)
```
