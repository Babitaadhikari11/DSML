# Accessing array elements

```python
# indexing and slicing
# 1D
a = np.array([2,3,4,5,6,7])
```

```python
a
```

```python
a[0]
```

```python
print(a[2::1])
```

```python
# reverse
print(a[::-1])
```

```python
# slicing and indexing in 2D
arr1 = np.random.randint(0, 10, size = (5,4))
```

```python
arr1
```

```python
arr1[2] #second index ko sabai row dincha
```

```python
arr1[2][-1] #second row ko  index 2 ko last element
```

```python
arr1[-1,-1]
```

```python
# last column access -  slicing 2D array
# array_name[row_idx1:row_idx2:step1, col_idx1:col_idx2:step2]
arr1[1:4:1, 2:]
```

```python
arr1[::, -2:] #all row last ko 2 ta column
```

```python
arr1[2:4, 1:3] 
```
