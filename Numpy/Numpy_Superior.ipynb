Why Numpy over python for loop - perform operation on entire arrays rather than individual elements

```python
import time
import numpy as np
```

```python
a = np.random.rand(90_000_000)
b = np.random.rand(90_000_000)
```

```python
# pure python using for multiplication

st = time.time()
res = []

for num1, num2 in zip(list(a), list(b)):
    res.append(num1 * num2)
    
end = time.time()
```

```python
print(f"time taken by python loop: {end -st}")
```

```python
# using list comprehension
st = time.time()


res = [num1 * num2 for num1, num2 in zip(list(a), list(b))]
    
end = time.time()
```

```python
print(f"time taken by list comprehension : {end -st}")
```

```python
# using numpy
st = time.time()
res = np.multiply(a, b)
end =  time.time()
```

```python
print(f"time taken by numpy  : {end -st}")
```
