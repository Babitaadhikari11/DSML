# Matrix wise operation

```python
# matrix multiplication
```

```python
a
```

```python
# transpose
b_transpose = np.transpose(b)
```

```python
b_transpose
```

```python
# multiplication
result = np.dot(a, b_transpose)
# a.dot(b_transpose)
```

```python
# trace -  diagnoal sum
d_sum = np.trace(result)
print(d_sum)
```
