## combining data from different sources into a single view

```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {"acc" : [1,2,3,4],
     "name": ["Ram", "Shyam", "Hari", "Ravi"],
     "age" : [21, 56,34,22]  
    }
)
```

```python
df2 = pd.DataFrame(
    {"acc" : [1,2,3,4],
     "principle": [500,677,300,3056],
     "interest" : [4.5,6, 7.5,3.1]  
    }
)
```

```python
df1
```

```python
df2
```

```python
# joining these two dataset
```

## Concatenation

```python
pd.concat([df1, df2], axis = 1) 
```

## Merging

```python
df1 = pd.DataFrame(
    {"key" : ["AB", "CB","AC", "DE"],
     "name": ["Ram", "Shyam", "Hari", "Ravi"],
     "age" : [21, 56,34,22]  
    }
)
df2 = pd.DataFrame(
    {"key" : ["AB", "CC","AC", "DF"],
     "principle": [500,677,300,3056],
     "interest" : [4.5,6, 7.5,3.1]  
    }
)
```

```python
df1, df2
```

```python
pd.merge(df1, df2, how = "inner",on = "key") #how =  inner, left, right,  outer joins, default = inner, on = key column
```

```python
pd.merge(df1, df2, how = "right",on = "key") 
```

```python
# alternative -  df1.merge(df2)
```
