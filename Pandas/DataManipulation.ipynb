## Handling Null value for age, using initials(Mr, Mrs, Miss or Master) groups

```python
titanic_df.initials.replace(
    to_replace= ["Rev", "Major", "Col", "Don", "Capt", "JonKheer"],
    value= "Mr", 
   )
titanic_df.initials.value_counts()
```

```python
titanic_df.initials = titanic_df.initials.replace(
    to_replace= {"Mile" : "Miss",
                 "Mme" : "Mrs",
                 "Ms" : "Mrs",
                 "Lady" : "Mrs",
                 "the Countess" : "Mrs"
                 }
)
```

```python
titanic_df.initials.value_counts()
```

```python
# convert dr to mr and ms for male and females
def replace_with_gender(row):
    if row.Sex == "male":
        value =  "Mr"
    else:
        value =  "Mrs"
    return row.initials.replace("Dr", value,  True)
```

```python
titanic_df.initials = titanic_df.apply(replace_with_gender, axis = 1)
```

```python
titanic_df.initials.value_counts()
```

```python
def get_initials(name):
    first_name = name.split(",")[-1]
    title = first_name.split(".")[0]
    return title.strip().lower()

titanic_df.Name.apply(get_initials)
```

### Outlier Handling - IQR AND Z-SCORE

# IQR - Inter Quartile Range

```python
titanic_df_fillAge_with_initial.Age.plot(kind = "box", vert = False )
```

```python
# for calculation IQR - We need Q1, Q2 and Q3
```

```python
Q1 = titanic_df_fillAge_with_initial.Age.quantile(q = 0.25)
Q3 = titanic_df_fillAge_with_initial.Age.quantile(q = 0.75)
IQR = Q3 - Q1
maximum = Q3 + (1.5 * IQR)
minimum = Q1 - (1.5 * IQR)
```

```python
# detecting outliers
Age_outlier = titanic_df_fillAge_with_initial[
    (titanic_df_fillAge_with_initial.Age > maximum) | (titanic_df_fillAge_with_initial.Age <minimum)
    ].shape
```

```python
Age_outlier
```

```python
# detecting inliers
Age_without_outliers = titanic_df_fillAge_with_initial[
    (titanic_df_fillAge_with_initial.Age <= maximum) | (titanic_df_fillAge_with_initial.Age >= minimum)
    ]
```

```python
# detecting outliers in Fare
Age_without_outliers.Fare.plot(kind = "box", vert = False, figsize = (12,2))
```

```python
Q1 = Age_without_outliers.Fare.quantile(q = 0.25)
Q3 = Age_without_outliers.Fare.quantile(q = 0.75)
IQR = Q3 - Q1
maximum = Q3 + (1.5 * IQR)
minimum = Q1 - (1.5 * IQR)
print(maximum, minimum)
```

```python
# detecting outliers for Fare
Fare_ = Age_without_outliers[
    (Age_without_outliers.Fare > maximum) | (Age_without_outliers.Fare <minimum)
    ]
```

```python
Fare_.shape
```

```python
Fare_.Fare.plot(kind = "box", vert = False, figsize = (12,2))
```

# Z-Score (mean bata kati tada) -  hypothesis being tested

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
x_axis =  np.linspace(-3,3,2000) # straight line
```

```python
y_axis = stats.norm.pdf(x_axis) #density curve
```

```python
plt.plot(x_axis, y_axis, "black", lw = 2, label = "Standard Normal Distribution")
```

# calculating and interpreting Z-Score

```python
sample = list(range(150,200, 4))
```

```python
sample
```

```python
sample_mean = np.mean(sample)
```

```python
sample_std = np.std(sample, ddof = 1)
```

```python
sample_mean, sample_std
```

```python
z_score = (sample - sample_mean)/ sample_std
```

```python
for i,z in enumerate (z_score):
    print(f"Data point {i}: {sample[i]} has z-Score = {z:.3f}")
```

```python
z_score = abs(
    (titanic_df_fillAge_with_initial.Age - titanic_df_fillAge_with_initial.Age.mean()) / titanic_df_fillAge_with_initial.Age.std()
)
```

```python
titanic_df_fillAge_with_initial = titanic_df_fillAge_with_initial[z_score <= 2.5]
```

```python
# save to csv
titanic_df_fillAge_with_initial.to_csv("clean_titanic_dataset.csv", index=False)
```
