a description for **each statistical factor**:

* **Count:** Age has **714 values**, meaning some ages are missing. Fare has **891 values**, so there are no missing fare values.
* **Mean:** The average passenger age is **29.7 years**, while the average fare is **32.20**.
* **Standard Deviation:** Age has a standard deviation of **14.53**, while Fare has **49.69**, showing that fares vary much more widely.
* **Minimum:** The youngest passenger was **0.42 years old**, and the minimum fare was **0**.
* **25% (Q1):** 25% of passengers were **20.13 years or younger**, and 25% paid **7.91 or less**.
* **50% (Median):** Half of the passengers were **28 years or younger**, and half paid **14.45 or less**.
* **75% (Q3):** 75% of passengers were **38 years or younger**, and 75% paid **31 or less**.
* **Maximum:** The oldest passenger was **80 years old**, while the highest fare was **512.33**.

The survival data shows that **61.6% of the passengers did not survive**, while **38.4% survived**. This indicates that **more passengers lost their lives than survived** the Titanic disaster.


```python
titanic_df.value_counts(["Sex", "Survived"], normalize = True)
```

The data shows that **52.5% of all passengers were males who did not survive**, while **12.2% were males who survived**. In comparison, **26.2% were females who survived**, while only **9.1% were females who did not survive**. This indicates that **female passengers had a much better survival outcome than male passengers**.

## Crosstab - Compute a simple cross tabulation of two (or more) factors.

```python
pd.crosstab(
    index = titanic_df.Sex,
    columns = titanic_df.Survived,
    margins = True,# gives the total number,default  =  All
    margins_name = "Total",
    colnames = ["Rate of survival"] #changing col name from surivied to rate of survival
)
```
