### Chi Square Test - show dependency between categorical variables

```python
import scipy.stats as stats
```

```python
# contengency table
contengency_table =  pd.crosstab(titanic_df.Pclass, titanic_df.Survived)
contengency_table
```

```python
chi2_score, pvalue, dof, expected_frequency = stats.chi2_contingency(contengency_table)
```

```python
chi2_score
```

```python
print(f"p value = {round(pvalue, 5)}") # lower than significance level (0.05) so alternate hypothesis is true
```

```python
dof
```

```python
expected_frequency
```
