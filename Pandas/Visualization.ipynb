# 1. Bar Chart

```python
titanic_df.Survived.value_counts()
```

```python
titanic_df.Survived.value_counts().plot(
    kind = "bar",
    title = "Count of Survived Passengers",
    figsize= (7,7), #agadi =  width, pachadi =  height
    legend= True, #color indication
    xlabel= "Survived (0 = No, 1 = Yes)",
    ylabel= "No. of Passengers",
    # grid= True
    yticks= list(range(0, 551, 50)), # y side ma number set gareko
    colormap= "Set2",#set3 ma different color huncha it uses from initial

)
```

```python
titanic_df.value_counts(subset= ["Sex", "Survived"])
```

```python
titanic_df.value_counts(subset= ["Sex", "Survived"]).plot(
    kind = "bar",
    title = "Count of Survived Passengers according to gender",
    figsize= (7,7), #agadi =  width, pachadi =  height
    legend= True, #color indication
    xlabel= "Survived (0 = No, 1 = Yes)",
    ylabel= "No. of Passengers",
    # grid= True
    yticks= list(range(0, 551, 50)), # y side ma number set gareko
    colormap= "Set2",#set3 ma different color huncha it uses from initial

)
```

```python
pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
)
```

```python
ax = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
).plot(
    kind = "bar",
    title = "Count of Survived Passengers according to gender",
    figsize= (7,7), #agadi =  width, pachadi =  height
    legend= True, #color indication
    xlabel= "Survived (0 = No, 1 = Yes)",
    ylabel= "No. of Passengers",
    # grid= True
    yticks= list(range(0, 551, 50)), # y side ma number set gareko
    colormap= "tab20",#set3 ma different color huncha it uses from initial

)
ax.bar_label(ax.containers[0], padding=2)
```

# 2. Pie Chart

```python
groupby_embarked = titanic_df.groupby(by= "Embarked")
groupby_embarked.count()
```

```python
groupby_embarked.count().plot(
    kind = "pie",
    y = "Survived",
    title="Survived passengers by pclass",
    colormap= "tab20",
    autopct = "%1.2f%%", #percentage inside piechart (1 =  integer and 2 indicates numbers to put after point)
    startangle = 90,
    figsize=(5,6)
)
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
)
crosstab
```

```python
crosstab.plot(
    kind="pie",
    subplots=True
)
```

# 8. Heat Map

```python
correlation =  titanic_df.corr(numeric_only=True)
correlation
```

```python
!pip install seaborn
```

```python
import seaborn as sns 
```

```python
import numpy as np
mask = np.triu(np.ones_like(correlation, dtype= bool)) #deviate the values to one side
```

```python
sns.heatmap(data = correlation, cmap = "coolwarm", fmt= ".2f", annot = True, vmin= -1, mask = mask ) #annot  gives number inside boxes
```

```python
# bubble chart using seaborn
sns.relplot(
    data =  titanic_df,
    x =  "Age",
    y = "Fare",
    size = "SibSp",
    hue = "Sex", #color distribution
    alpha = 0.75 #transparent
)
```

```python
# violin plot
sns.violinplot(
    data = titanic_df,
    y = "Age",
    x= "Sex",
    hue = "Survived",
    inner="quart",
    split = True,
    orient = "v"
   
    
)
```
