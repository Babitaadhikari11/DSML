```python
# class

class Calculator:

    # initializer(constructor)
    def __init__(canbeanything, num1, num2): #object aayera yaa bascha
        print("making object..")
        canbeanything.num1 =  num1
        canbeanything.num2 = num2

    # functions,method
    def addition(canbeanything): #calc object ko property bhanera define gareko
        return canbeanything.num1 + canbeanything.num2
    
    def multiply(canbeanything, num3: int) -> int | float:
        """
        Documentation
        """
        return canbeanything.num1 * canbeanything.num2 * num3
    
```

```python
# object
calc = Calculator(4,4)
calc.addition()
```

```python
calc.multiply(2)
```

```python
class Dog:
    # class attribute, not associated with objects
    count = 0

    # iniitializer
    def __init__(self, name:str, age: int = 1):
        self.name = name #instance attribute, changeable
        self.age = age

        Dog.count += 1
```

```python
# object
dog1 = Dog("Luffy", 3)
dog1.name
```

```python
dog1.count
```

```python
# object, instance objects are changed here
dog2 = Dog("Max", 3)
dog2.name, dog2.age
```

```python
dog2.count #everytime object is created, count value is updated
```

Methods Type (Instance, Class, Static)

```python
class Person:
    count = 0 #class attribute
    
    def __init__(self, name: str, address: str ="Kathmandu"):
        self.name =  name #instance attributes
        self.address =  address

        Person.count += 1

    # instance method
    def get_details(self): #instane method, object as an parameter
        print(f"i am {self.name} from {self.address}")
        
    # class method
    @classmethod #decorator
    def get_count(cls): #variable can be differently used except used variable in initializer(object)
        return cls.count
    
    # static method
    @staticmethod
    def get_full_name(first_name: str, last_name: str):
        return f"{first_name}, {last_name}"
```

```python
object1 =  Person("Babita", "Lalitpur")
object1.get_details()
```

```python
object1.get_count()
```

```python
object1.get_full_name("Ram", "Rai")
```

```python
#  class to accept student name and marks in 3 subjects , calculate and display total and average marks
class Student:

    def __init__(self, name, subj1, subj2, subj3):
        self.name = name
        self.sub1 = subj1
        self.sub2 = subj2
        self.sub3 = subj3
    
    def total(self):
        total_marks = self.sub1 + self.sub2 + self.sub3
        return total_marks
    
    def average(self):
        total_ = self.total()
        average = total_ / 3
        return average

    def display(self):
        return f"student with name {self.name} obtained total marks {self.total()} and average is {self.average()}"
    

        
        
```

```python
student1 = Student("Ram", 20, 30, 40)
student1.display()
```

```python
student2 = Student("Shyam", 22, 33, 44)
student2.display()
```

```python
student3 = Student("Sita", 100, 100, 100)
student3.display()
```
