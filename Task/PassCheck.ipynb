```python
class PassCheck:

    def __init__(self, password : str):
        self.password =  password
    
    def __str__(self):
        return f"{self.password[0]}***{self.password[-1]}"
    def __call__(self):
        if len(self.password) > 8 and not self.password.isalnum():
            return "strong"
        
        return "weak"
        
    # calling by obj
```
