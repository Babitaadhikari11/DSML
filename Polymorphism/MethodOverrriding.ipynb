Practice Question — Method Overriding

Create a parent class called Vehicle with:

A method start() that prints "Vehicle is starting"

Create two child classes:

Car
Bike

Both classes should override the start() method.

```python
# parent class
class Vehicle:

    def start(self):
        print("Vehicle is starting")
```

```python
# child class 1
class Car(Vehicle):

    # overridde start
    def start(self):
        print("Car is starting")
```

```python
# child class 2
class Bike(Vehicle):

    # override 
    def start(self):
        print("Bike is starting")
```

```python
# calling
bike = Bike()
```

```python
bike.start()
```

```python
car = Car()
```

```python
car.start()
```
