```python
# a single method can have arguments with different types
class Example:

    def add(self, a, b, c = 0):
        return a + b + c
```

```python
obj = Example()
obj.add(2,3) 
```

```python
# using variable length arguments
class Example:

    def add(self, *args):
        return sum(args)
```

```python
obj1 = Example()
print(obj1.add(12,3,4,5))
```
