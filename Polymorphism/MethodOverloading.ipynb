```python
# a single method can have arguments with different types
class Example:

    def add(self, a, b, c = 0):
        return a + b + c
```

```python
obj = Example()
obj.add(2,3) 
```

```python
# using variable length arguments
class Example:

    def add(self, *args):
        return sum(args)
```

```python
obj1 = Example()
print(obj1.add(12,3,4,5))
```

Practice Question — Method Overloading

Create a class called Calculator.

Create a method called add() that can work with:

2 numbers → return their sum
3 numbers → return their sum

```python
class Calculator:

    def add(self, a, b, c = None):
        if c is not  None:
            return a + b + c
        return a + b
```

```python
calc = Calculator()
```

```python
# not c
calc.add(1,1)
```

```python
# there is c
calc.add(1, 1, 2)
```
