```python
import pandas as pd
```

```python
student_df =  pd.read_csv("cleaned_student_dataset.csv")
```

```python
student_df.head()
```

## Grouping Categories

```python
student_df.faculty.value_counts()
```

```python
faculty_group = {
    "Science": "STEM",
    "Engineering": "STEM",
    "AgriSciences": "STEM",
    "Medicine and Health Services": "STEM",
    "Economic & Management Sciences": "Management",
    "Arts & Social Sciences": "Others",
    "Law": "Others",
    "Education": "Others"
}
```

```python
student_df["faculty"] = student_df["faculty"].replace(faculty_group)
```

```python
student_df["faculty"].value_counts()
```

```python
# FOR gender
student_df.gender.value_counts()
```

```python
# for study_year
student_df.study_year.value_counts()
```

```python
study_year_mapping = {
    "Highschool": "Junior",
    "1st Year": "Junior",
    "2nd Year": "Mid",
    "3rd Year": "Mid",
    "4th Year": "Senior",
    "Postgraduate": "Senior"
}
```

```python
student_df["study_year"] = student_df["study_year"].map(study_year_mapping)
```

```python
student_df.study_year.value_counts()
```

```python
student_df
```

```python
student_df["accomodation"] = student_df["accomodation"].replace({
    "Private accommodation/ stay with family/friends" : "Private",
    "Non-private accommodation ie. Res " : "Public",
    "Unknown" :"Unknown"
    
})
```

```python
student_df.accomodation.value_counts()
```

```python
# for allowance
student_df.allowance.value_counts()
```

```python
allowance_map = {
    "Unknown": 5,
    "R 4001- R 5000": 0,
    "R 5001 - R 6000": 1,
    "R 6001 - R 7000": 2,
    "R 7001 - R 8000": 3,
    "R 8000+": 4
}
```

```python
student_df["allowance"] = student_df["allowance"].map(allowance_map)
```

```python
student_df.allowance.value_counts()
```

```python
# for scholarship
student_df.scholarship.value_counts()
```

```python
student_df["scholarship"] = student_df["scholarship"].replace({
    "Yes (NSFAS, etc...)" : 0,
    "No" : 1  ,
    "Unknown" : 2
})
```

```python
student_df.scholarship.value_counts()
```

```python
# for study hour
student_df.study_hours.value_counts()
```

```python
study_hours_ = {
    "0": 0,
    "1-3": 1,
    "3-5": 2,
    "5-8": 3,
    "8+": 4
}

student_df["study_hours"] = student_df["study_hours"].replace(study_hours_)
```

```python
student_df.study_hours.value_counts()
```

```python
# for party week
student_df.party_week.value_counts()
```

```python
student_df["party_week"] = student_df["party_week"].replace({
    "0": 0,
    "1": 1,
    "Only weekends": 2,
    "2": 3,
    "3": 4,
    "4+": 5
})
```

```python
student_df.party_week.value_counts()
```

```python
# DRINKS NIGHT
student_df.drinks_night.value_counts()
```

```python

student_df["drinks_night"] = student_df["drinks_night"].replace({
     "0": 0,
    "1-3": 1,
    "3-5": 2,
    "5-8": 3,
    "8+": 4
})
```

```python
student_df.drinks_night.value_counts()
```

```python
# missed class
student_df.missed_classes.value_counts()
```

```python
student_df["missed_classes"] = student_df["missed_classes"].replace({
     "0": 0,
    "1": 1,
    "2": 2,
    "3": 3,
    "4+": 4
})
```

```python
student_df.missed_classes.value_counts()
```

```python
# failed class
student_df.failed_classes.value_counts()
```

```python
student_df["failed_classes"] = student_df["failed_classes"].replace({
     "0": 0,
    "1": 1,
    "2": 2,
    "3": 3,
    "4+": 4
})
```

```python
student_df.failed_classes.value_counts()
```

```python
# inrelationship
student_df.in_relationship.value_counts()
```

```python
student_df["in_relationship"] = student_df["in_relationship"].replace({
    "Yes" : 0,
     "No": 1
})
```

```python
student_df.in_relationship.value_counts()
```

```python
# parent aprove
student_df["parent_approve"] = student_df["parent_approve"].replace({
    "Yes" : 0,
     "No": 1
})
```

```python
student_df.parent_approve.value_counts()
```

```python
# relation with parent
student_df.relation_parent.value_counts()
```

```python
student_df["relation_parent"] =  student_df["relation_parent"].replace({
    "Distant": 0,
    "Fair": 1,
    "Close": 2,
    "Very close": 3
})
```

```python
student_df.relation_parent.value_counts()
```

```python
student_df.faculty.value_counts()
```
