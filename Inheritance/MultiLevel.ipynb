```python
# multilevel inheritance
class A: 

    def method1(self):
        print("this is method from class A")

class B(A): 

    def method2(self):
        print("this is method from class B")

class C(B): 

    def method3(self):
        print("this is method from class C")
```

```python
obj= C()
obj.method1()
```

```python
obj.method2()
```

```python
obj.method3()
```

```python
# 
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age
        
```

```python
class Employee(Person):

    def __init__(self, name, age, empID):
        super().__init__(name, age)
        self.empID = empID
```

```python
class Manager(Employee):

    def __init__(self, name, age, empID, department):
         super().__init__(name, age, empID)
         self.department = department

    def get_details(self):
        print(f"Name: {self.name} of age {self.age} with employee ID {self.empID} is in {self.department} deparment")
```

```python
manager = Manager("Babita", 21, 101, "Lab")
manager.get_details()
```

hierarchical inheritance, ONE base class and multiple derived class

```python
# base class
class Animal:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"Hello, i am {self.name}")
```

```python
# derive class 1
class Cat(Animal):
    
    def speak(self):
        return "meow"
```

```python
# derive class 2
class Dog(Animal):
    
    def speak(self):
        return "woof"
```

```python
# derive class 3
class Lion(Animal):
    
    def speak(self):
        return "roar"
```

```python
cat = Cat("Max", 4)
```

```python
print(cat.intro()) 
```

```python
print(cat.speak())
```

Hybrid Inheritance - two or more types of inheritance combination

```python
# base class
class Animal:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"Hello, i am {self.name}")
```

```python
class Mammal(Animal):

    def __init__(self, name, age, fur = False):
        super().__init__(name, age)
        self.fur = fur

        def birth(self):
            print(f"{self.name} gives birth to live ones")
        
```

```python
class Birds(Animal):

    def fly(self):
        return f"{self.name} can fly"
```

```python
class Bat(Mammal, Birds): #combination of hierarchical and multiple inheritance = hybrid
    pass
```

```python
obj = Bat("Help", 3)
```

```python
obj.intro()
```

```python
obj.fly()
```
