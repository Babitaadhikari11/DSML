multiple parents


```python
class Parent1:
    # here is default initializer
    
    # method 1
    def function1(self):
        print("this is parent 1")
```

```python
class Parent2:
    
    # method 2
    def function2(self):
        print("this is parent 2")
```

```python
class Parent3:
    
    # method 3
    def function1(self):
        print("this is parent 3")
```

```python
# derived class
class ChildClass(Parent3, Parent2, Parent1):

    def function4(self):
        print("This is child function")
```

```python
# object
obj = ChildClass()
obj.function1()
```

```python
obj.function4()
```

```python
obj.function2()
```
