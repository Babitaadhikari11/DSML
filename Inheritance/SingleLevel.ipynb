```python
# single inheritance
# parent, base class
class Animal:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"Hello, i am {self.name}")
```

```python
# sub class
class Cat(Animal): #child class accessing base class with inherting its attributes  

    def __init__(self, name, age, color): #this initializer overrides parent's initializer

        # accessing parent initializer
        super().__init__(name, age)
        self.color = color

    def details(self):
        print(f"{self.color} cat: \n {self.name} \nage: {self.age}")
```

```python
# class cat object
cat = Cat("Meow", 2, "Brown")
cat.intro()
cat.details()
```
