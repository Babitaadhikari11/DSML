## Filtering and Sorting:
#### Filter the dataset to show only the rows where the sepal_length is greater than 5.0.
#### Sort the filtered dataset by petal_length in descending order.
 

```python
greater_sepal_length = iris_df[iris_df["SepalLengthCm"] > 5.0]
```

```python
greater_sepal_length.head()
```

```python
greater_sepal_length.shape
```

The filtered dataset shows only the records where **SepalLengthCm is greater than 5.0 cm** They are 118 out of 150. The results include flowers from all three species: Iris-setosa, Iris-versicolor, and Iris-virginica. Records with a sepal length of **5.0 cm or less are excluded**.

```python
sorted_petal_length = greater_sepal_length.sort_values(
    by= "PetalLengthCm",
    ascending= False
    
)
```

```python
sorted_petal_length
```

The filtered dataset is sorted by **PetalLengthCm in descending order**, from the largest to the smallest petal length. The highest petal length is **6.9 cm (Iris-virginica)**, while the lowest in the filtered data is **1.2 cm (Iris-setosa)**.

## 5. Creating New Columns:
#### Create a new column called sepal_petal_ratio which is the ratio of sepal_length to petal_length.

```python
iris_df["sepal_petal_ratio"] = iris_df.SepalLengthCm / iris_df.PetalLengthCm
```

```python
iris_df.head()
```

A new column `sepal_petal_ratio` was created by dividing **SepalLengthCm by PetalLengthCm**. It shows the ratio between the sepal length and petal length for each flower.

### 6. Data Visualization:
#### Histogram:
##### Create a histogram to visualize the distribution of sepal_length values.
#### Scatter Plot:
##### Create a scatter plot with sepal_length on the x-axis and petal_length on the y-axis. Use different colors to represent different species.
#### Box Plot:
##### Create a box plot for the sepal_width values, grouped by species.

```python
import matplotlib.pyplot as plt

iris_df["SepalLengthCm"].plot(
    kind = "hist",
    bins = 10,
    edgecolor="black",
    xlabel = "Sepal length in cm"
)
```

The histogram shows the distribution of **sepal length** in the Iris dataset. Most flowers have sepal lengths between approximately **5.0 cm and 6.5 cm**, while fewer flowers have very small or very large sepal lengths.

```python
import seaborn as sns

sns.scatterplot(
    data = iris_df,
    x = "SepalLengthCm",
    y = "PetalLengthCm",
    hue="Species",
)
plt.xlabel("Sepal Length in cm")
plt.ylabel("Petal Length in cm")
plt.title("Sepal Length vs Petal Length")

plt.show()
```

The scatter plot shows the relationship between **sepal length and petal length** for the three Iris species. Iris-setosa has the shortest petals, Iris-versicolor has medium-sized petals, and Iris-virginica generally has the longest petals. Overall, the plot shows that **petal length tends to increase as sepal length increases**.

```python
sns.boxplot(
    data=iris_df,
    x= "Species",
    y= "SepalWidthCm",

)
```

The box plot compares **sepal width across the three Iris species**. **Iris-setosa has the highest median sepal width**, Iris-versicolor has the lowest, while Iris-virginica falls in between. The plot also shows a few **outliers for Iris-virginica**, indicating unusually high or low sepal width values.


### Correlation Matrix
#### Calculate the correlation matrix for the numerical features and create a heatmap to visualize the correlations using seaborn.

```python
heat_map=iris_df.corr(numeric_only= True)
sns.heatmap(
    heat_map,
    annot= True,
    cmap="viridis"
)
```

The heatmap shows the correlation between the numerical features. **PetalLengthCm and PetalWidthCm have the strongest positive correlation (0.96)**, meaning they tend to increase together. **SepalLengthCm also has a strong positive correlation with PetalLengthCm (0.87)**. In contrast, **SepalWidthCm has weak to moderate negative correlations with most flower measurements**.
