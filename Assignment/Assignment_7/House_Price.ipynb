## Problem 3: Handling Missing Data and Imputation

### 1. Identify Missing Values
Check the dataset to identify columns containing missing values.

### 2. Impute Missing Values
- Replace missing values in **numerical columns** with the **median**.
- Replace missing values in **categorical columns** with the **most frequent value (mode)**.

### 3. Justify your choice of imputation methods and check if any missing data remains.

```python
for i in house_df.columns:
    if house_df[i].isna().sum() >0:
        print(i)
        
```

```python
for i in house_df.select_dtypes(include="number").columns:
    house_df[i] = house_df[i].fillna(house_df[i].median())
```

```python
for i in house_df.select_dtypes(include="str").columns:
    house_df[i] = house_df[i].fillna(house_df[i].mode()[0])
```

```python
house_df.isna().sum()
```

The missing values were filled according to their data types. For **numerical columns**, missing values were replaced with the **median**, as it is less affected by outliers. For **categorical columns**, missing values were replaced with the **most frequent value (mode)**. After imputation, all columns contain **0 missing values**, confirming that no missing data remains.

1. Perform a **groupby** operation on `Neighborhood` and calculate the **mean** and **median** for `SalePrice`.

2. Use the `apply()` function to create a custom transformation for `SalePrice` (e.g., scaling or transformation).

3. Aggregate multiple statistical values (e.g., **mean**, **median**, and **count**) for `SalePrice` grouped by `OverallQual`.

```python
house_df.groupby("Neighborhood")["SalePrice"].aggregate(["mean","median"])
```

The data is grouped by Neighborhood and the mean and median SalePrice are calculated for each area. NoRidge has the highest mean sale price of approximately 335,295, while MeadowV has the lowest at approximately 98,576.

```python
min_val = house_df["SalePrice"].min()
max_val = house_df["SalePrice"].max()
def scale_price(x):
    return (x - min_val) / (max_val - min_val)
```

```python
house_df["Scaled_Sale_Price"] = house_df["SalePrice"].apply(scale_price)
```
