```python
tags_input=""" python , data-science,
machine-learning"""
```

```python
# split string into a list of strings by comma
raw_list=tags_input.split(",")



```

```python
# extract only first tag and clean surrounding whitespace
first_tag=raw_list[0].replace(" ","")
first_tag
```

```python
# replace hyphens with spaces in second tag
second_tag_clean=raw_list[1].strip().replace("-"," ")
second_tag_clean
```

```python
# create new list
cleaned_tags=[first_tag, second_tag_clean]
```

```python
# adding last tag cleaned to end of cleaned_tags
last_tag_clean=raw_list[-1].strip()
cleaned_tags.append(last_tag_clean)
cleaned_tags
```
