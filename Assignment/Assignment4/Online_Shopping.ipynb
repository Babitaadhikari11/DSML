```python
# step 2
class User:

    def __init__(self, name, premium_member =  False, is_admin =  False):
        self.name = name
        self.premium_member = premium_member
        self.is_admin = is_admin
    
    def make_premium(self, other_user):
        if self.is_admin:
            other_user.premium_member = True
        else:
            print("Only admin can make user premium")

    def make_admin(self, other_user):
        if self.is_admin:
            other_user.is_admin = True
        else:
            print("Only admin can create another admin")
```

```python
user1 = User("Babita", True)
```

```python
user2 = User("Riya", False)
```

```python
print(user1.name)
print(user1.premium_member)
```

```python
class ShoppingCart:

    def __init__(self, user):
        self.user = user
        self.cart = []

    def add_product(self, product):
        self.cart.append(product)

    def remove_product(self, product):
        self.cart.remove(product)

    def calculate_total_cost(self):
        total = 0

        for product in self.cart:
            total += product.price

        if self.user.premium_member:
            total = total - (total * 0.10)

        return total
    
    def generate_invoice(self):

        print("Invoice")

        for product in self.cart:
            print(product.name, "-", product.price)

        print("Total:", self.calculate_total_cost())
```
