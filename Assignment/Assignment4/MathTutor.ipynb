### Math Tutor Using Random, Arithmetic Operators, and OOP

### Scenario:
Create a MathTutor class that generates random math questions using random and math operators

### Track correct and incorrect answers.

### Provide a score at the end.

### Handle invalid input using exception handling.

### Concepts:
Random, Math, Exception Handling, Classes

```python
import random
import math
```

```python
import random

class MathTutor:

    def __init__(self):
        self.correct = 0
        self.incorrect = 0

    def ask_question(self):

        num1 = random.randint(1, 10)
        num2 = random.randint(1, 10)

        operators = ["+", "-", "*", "/"]
        operation = random.choice(operators)

        # calculate answer
        if operation == "+":
            result = num1 + num2

        elif operation == "-":
            result = num1 - num2

        elif operation == "*":
            result = num1 * num2

        else:
            result = num1 / num2


        try:
            print("What is", num1, operation, num2, "?")

            answer = float(input("Enter your answer: "))

            if answer == result:
                print("Correct!")
                self.correct += 1

            else:
                print("Incorrect!")
                print("Correct answer:", result)
                self.incorrect += 1

        except ValueError:
            print("Please enter a valid number")
            self.incorrect += 1

    def show_score(self):

        print("Correct:", self.correct)
        print("Incorrect:", self.incorrect)
        print("Score:", self.correct)

```

```python
# object
math = MathTutor()

```

```python

for i in range(5):
    math.ask_question()
```

```python

math.show_score()
```
